```python
import os
import sys
import urllib.request
import zipfile
```

```python
COCO_BASE_URL = "http://images.cocodataset.org"
FILES = {
    "train_images": {
        "url": f"{COCO_BASE_URL}/zips/train2017.zip",
        "filename": "train2017.zip",
    },
    "annotations": {
        "url": f"{COCO_BASE_URL}/annotations/annotations_trainval2017.zip",
        "filename": "annotations_trainval2017.zip",
    },
}
```

```python
root_dir = "../../coco2017"
```

```python
def download_file(url, dest_path):
    print(f"Downloading: {url}")
    urllib.request.urlretrieve(url, dest_path)
    print(f"Saved to: {dest_path}")

def extract_zip(zip_path, extract_to):
    print(f"Extracting: {zip_path}")
    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(extract_to)
    print(f"Extracted to: {extract_to}")

os.makedirs(root_dir, exist_ok=True)

for key, file_info in FILES.items():
    zip_path = os.path.join(root_dir, file_info["filename"])

    download_file(file_info["url"], zip_path)
    extract_zip(zip_path, root_dir)
```
